Gupta et al. 2024

## 1. Introduction

### Notes

* False information is spread with or without malicious intent
* Hard to do just by humans
* Google and Yelp have fake review detection algorithms but they are not publicly accessible
* Brand marketers purchase fake positive reviews from both unkonwn and unkonwn reviewers
* May employ fake negative reviews to maliciously damage the reputation of their competitors

## 2. Overview of the Problem

### Notes

* Financial incentives: Reviewers may receive direct payments, or other benefits for writing a fake review
* Endorsement help: helping out a friend or boosting self-esteem
* Reputation management: Certain businesses may make certain positive reviews to counteract negative ones
* Posting negative reviews to get product for free
* Seeking revenge
* Competitor sabotage
* Cold start of products or services
* Competition to maintain competitiveness in the market
* disruptive spam reviews that offer no relevant opinions or insights regarding the product at hand

### Types of reviews

Jindal and Liu 2007
* Type I: deceptive reviews
* Type II: brand-specific reviews
* Type III: disruptive reviews or non-reviews

Salimen et al.
* OR: Original reviews
* CG: Computer generated reviews


### How is a fake review detected?

1. Data Collection
* dataset is compiled containing a large number of reviews, genuine and fake ones. Can be sourced from various platforms such as e-commerce websites, social media, or review aggregation sites
2. Feature Extraction
* Relevant features are extracted from reviews
* textual information like the review text, user profile details, timestamps, ratings, other metadata
* sentiment analysis, length of review, language patterns
3. Data preprocessing
* Collected data is preprocessed to ensure consistency and improve the quality of input
* removing noise, normalizing text
* (lowercasing, removing punctuation), handling missing data, transforming features into a suitable format for AI algorithms
4. Labelling the data
* Each review in the dataset needs to be labelled as either genuine or fake.
* this can be done manually by human reviewers who are familiar with fake review patterns
* Or by using existing labelling techniques such as rule-based, algorithm-based, crowdsourced labelling (Mewada & Dewang 2022)
5. Model Training
* Logistic regression, decision trees, random forests, svm, RNN, CNN, etc
6. Model Evaluation
* Trained model is evaluated using different evaluation metrics such as accuracy, precision, recall, and F1-score metrics
7. Model Deployment

## 3. Literature review

### Most used ML techniques

Supeverised ML algos
* SVM, KNN, LR are most commonly used methods (Tufail et al 2022)
    * Used on Yelp hotel review dataset
* Naive Bayes, Eristic Regression and SVM (Kumaran et al. 2021)
    * Dictionary based on social media keywords and online reviews
    * very limited feature set of uni-, bigrams and length of review
* SVM for fake review detection (Poonguzhali et al. 2022)
    * online e-commerce interface
    * Fake reviews are predicted in this system and are not used in the db while recommending a product on this interface
    * Method can't upload more than one review by one user 
    * Prevention method is not feasible in large databases

Semi-supervised ML Algorithms
* semi-supervised approach uses a pre-defined set of features to train classifiers
* Jing-Yu et al. (2022)
    * AspamGAN
    * Uses attention mechanism in the classifier to detect fake reviews
    * Used the data from the TripAdvisor dataset
    * Better performance with limited label data than SpamGAN (Stanton 2020)
* Lighthart et al. (2021)
    * four semi-supervised techniques
    * training on the labelled portion of the data
    * co-training (utilizing additional perspectives of data)
    * Transductive SVM (variation of traditional SVM used in s. supervised learning)
    * label propagation plus spreading
    * Train on one dataset and perform experiments on two more datasets from Yelp
    * Have not considered metadata or reviewer-based features
* Wu et al. (2022)
    * Semi-supervised probabilistic ensemble that collectively captures the individual behavioural characteristics of reviewers as well as the reviewer network
    * 10 behavioural features such as the number of reviews in a day, burstiness, popularity of product that the user reviewed, average distance between a user and other users
    * Assume that the reviewer network presents homophily

Unsupervised ML Algorithms
* Mothukuri et al. (2022)
    * Creates clusters using extracted features
    * K-means clustering, GMM Full covariance clustering, GMM Diagonal covariance unsupervised techniques to detect fake reviews taken from cafe dataset of Yelp
    * Found that K-means shows the highest accuracy among the three
    

### Deep Learning Techniques

Commonly used Deep Learning Algos for FRD:
* CNN, RNN, Transformer=based Models, Deep Belief Networks, Siamese Neural Netowrks
* Alsubari et al. (2021)
    * used four datasets including hotels, restaurants, Yelp, Amazon
    * Performed convolutional and max-pooling layers of the CNN to reduce dimensions and extract features
    * Cross-domain analysis
    * performance of this model decreases on a single-domain dataset
* Basyar et al. (2020)
    * LSTM and Gated Recurrent Unit model to detect e-mail spam
    * LSTM was slightly better but not significant
* Liu et al. (2022)
    * layered attention network employing two stratums to capture semantic information.
    * authors then integrate a Convolutional structure and Bi-LSTM to extract crucial semantics
    * Resulted in superior performance compared to other algorithms
    * smaller number of extracted features
* Crawford et al (2021)
    * Deep learning algo to predict the star ratings that will match the review
    * Don't predict whether the review is fake or real
* Salminene et al. (2022)
    * Fine-tuned the RoBERTa model, which is an optimized model of the BERT transformer
    * called it fakeRoBERTa
* Deep Learning algorithms outperform traditional machine learning models when applied to large-scale datasets
    * deep learning models are computationally expensive
    * Deep learning models are susceptible to overfitting and thus can't be used for smaller data
    * Transformer based models can be uninterpretable
    

### Reviewer-centric models

* Tufail et al. (2022)
    * review time, writing style, relationship words, grammatical errors, punctuation
* Alsubari et al. (2022)
    * 90% of genuine reviewers usually write one review a day, based on the product they bought or the service they used
    * 70% of fraudsters may write up to 5 reviews a day
    * Dataset is limited to hotel domain
* Sadiq et al.
    * The reviewers tend to give a higher proportion of positive fake reviews compared to negative fake reviews
    * Rating of a fake reviewer tends to be different from the ratings of the genuine reviewer
    * Users with longer profile timelines tend to be more authentic than the users who have recently created their profiles
    * Difference between the ratings given and the emotional intensity of the review is also a giveaway
* Computational cost is less as compared to review-centric feature selection


### Review-centric models

* Noun verb and adjective count along with the usage of strong positive and negative words in the review
* writing style of reviewer can have lexical characters 
    * number of characters and their proportion to uppercase letters
    * numeric characters,
    * rate of spaces or tabs
    * stylometric features 
* Linguistic Inquiry and Word Count
    * emotions
    * self reference
    * social words
    * overall cognitive words
* Moon et al. (2021)
    * Lack of details, temporal bias, hyperbole are all part of fraudulent reviews
    * Used hotel reviews
* Kumaran et al. (2021)
    * language features like unigrams and frequency
    * bigrams and frequency, 
    * length of review
* Abdulqader et al. (2022)
    * short length of the online review
    * review replication
    * Term frequency-inverse document frequency
    * cohesion and coherence measures
    * stylometric features
    * less usage of personal pronouns
    * less information about time and location
    * strong use of positives and negative words
* Hussain et al. (2020)
    * text similarity is widely used as an indication of fake reviews
* Wang and Kuan (2022)
    * message level (review length and psychological cues)
    * formulation level (readability and linguistic variables)
    * control level (rating extremity, review age, etc.)
    

### Important behavioural and linguistic pointers to identify fake reviews.

Hussain et al. (2020)
* Linguistic and behavioural pointers
    * Absense of profile data of the reviewer
    * Posting duplicate reviews
    * short and often grammatically erroneous reviews
    * groups of reviews with the same timestamp
    * excessive use of positive or negative words
* Only studied behavioural model and linguistic model separately
* Chances of reviews being classified accurately decrease when only one model is used.

Alsubari et al. (2022)
* Extracted features
    * sentiment score
    * four grams
    * POS counts
    * strong postive and negative word counts
* Models used
    * SVM, Naive Bayes, Random Forest, AdaBoost
* Limitations
    * Fewer extracted features
    * limited to hotel domain

Alawadh et al. (2023a)
* Experiments
    * balanced hotel review dataset
    * real-time application of deep learning-based, semantically aware text features
    * Used a small dataset which doesn't fully utilized NN advantage

Kumar et al. (2022)
* Novel ML framework based on M-SMOTE addresses the class imbalance problem. 
    * Combining reviewer-centric features with review-centric features significantly improves the performance of FRD models
* Limitations
    * features extracted in this dataset may not be present in other domains

Theuerkauf and Peters (2023)
* labelled reviews source from the iOS App store
