# Project Proposal: Fraudulent Review Detection in Apartment Reviews in Orem and Provo UT

## Problem Statement:

Online reviews can provide good insight on the quality of a product or service. They can help give a potential customer a more accurate view of the experience of a real person with the product or experience. However, in Orem and Provo Utah, students seeking housing are running into a problem. There have been reports that the reviews that are on google maps or other review websites may tend to be innaccurate or skewed, giving students a harder time understanding what the real experience is living at a certain apartment. The primary suspicion is that the apartment complexes are providing themselves fake reviews or having their employees write positive reviews to boost the perception of their apartment complex in a positive light.

## Solution

A model that can accurately take information about a review or a distribution of reviews and provide information about whether or not the review is deceptive or not.
